The TISE in spherical coordinates is given as:
$$
    -\frac{\hbar^2}{2m} \left( \frac{1}{r^2} \frac{\partial}{\partial r} \left( r^2 \frac{\partial \Psi}{\partial r} \right) + \frac{1}{r^2 \sin \theta} \frac{\partial}{\partial \theta} \left( \sin \theta \frac{\partial \Psi}{\partial \theta} \right) + \frac{1}{r^2 \sin^2 \theta} \frac{\partial^2 \Psi}{\partial \phi^2} \right) + V \Psi = i \hbar \frac{\partial \Psi}{\partial t}
$$
Solving this by separation of variables and imposing the conditions as per the hydrogen atom, we get spherically symmetric solutions as described below:

The general hydrogen wave function in spherical coordinates is given as:
$$
 \psi_{n\ell m}(r, \theta, \phi) = R_{n\ell}(r) Y_{\ell m}(\theta, \phi)
$$
where
$$
    R_{nl}(r) = \sqrt{\left( \frac{2}{na_0} \right)^3 \frac{(n-l-1)!}{2n[(n+l)!]^3}} \left( \frac{2r}{na_0} \right)^l e^{-\frac{r}{na_0}} L_{n-l-1}^{2l+1} \left( \frac{2r}{na_0} \right) = \text{Radial part}
$$ and
$$
  Y_{\ell}^{m}(\theta, \phi) = \sqrt{\frac{(2\ell+1)}{4\pi} \frac{(\ell-m)!}{(\ell+m)!}} P_{\ell}^{m}(\cos \theta) e^{im\phi} = \text{Angular part}
$$
Here, $\theta \in [0,\pi]$ is the Polar angle and $\phi \in [0,2\pi]$ is the Azimuthal angle. $P_{l}^{m}(x)$ and $L^{k}_{j}(x)$ are the associated Legendre and Laguerre polynomials respectively.

## 2
The radial function gives a visualisation of how the electron clouds are distributed in the orbitals, from the nucleus. They are determined by "n" (Principal Quantum Number) and "l" (Azimuthal/(Orbital) Angular Momentum Quantum Number - which determines shape of the orbital, as stated previously). <br>
The radial *distribution* function is given as:
$$
D_{nl}(r) = 4\pi r^{2}[R_{nl}(r)]^{2}
$$
We shall plot the radial distribution function for some orbitals and principal quantum numbers

#### (a) Importing necessary packages

```python
import matplotlib.pyplot as plt
import numpy as np
from sympy.abc import n,l,r,Z
from sympy.physics import hydrogen
from sympy.utilities.lambdify import lambdify
```

#### (b) Plotting some radial distribution functions

```python
R_nl_1 = lambdify(r,hydrogen.R_nl(1,0,r,1)) #1s
R_nl_2 = lambdify(r,hydrogen.R_nl(2,0,r,1)) #2s
R_nl_3 = lambdify(r,hydrogen.R_nl(2,1,r,1)) #2p
R_nl_4 = lambdify(r,hydrogen.R_nl(3,1,r,1)) #3p
R_nl_5 = lambdify(r,hydrogen.R_nl(3,2,r,1)) #3d
R_nl_6 = lambdify(r,hydrogen.R_nl(4,2,r,1)) #4d
R_nl_7 = lambdify(r,hydrogen.R_nl(4,3,r,1)) #4f
R_nl_8 = lambdify(r,hydrogen.R_nl(5,3,r,1)) #5f

r = np.linspace(0,150,1000)

fig,((ax1,ax2),(ax3,ax4),(ax5,ax6),(ax7,ax8)) = plt.subplots(4,2,figsize=(12,7))

plt.suptitle("Radial Distribution Function for some orbitals",fontsize=18)
fig.tight_layout(pad=3.5)

ax1.plot(r,4*np.pi*(r**2)*(R_nl_1(r)**2),label="1s")
ax1.set_title("1s",fontweight='bold')
ax1.set_xlim(0,70)
ax1.set_ylim(0,max(4*np.pi*(r**2)*(R_nl_1(r)**2)))
ax1.set_xlabel("r")
ax1.set_ylabel("$D_{nl}(r)$")

ax2.plot(r,4*np.pi*(r**2)*(R_nl_2(r)**2),label="2s")
ax2.set_title("2s",fontweight='bold')
ax2.set_xlim(0,70)
ax2.set_ylim(0,max(4*np.pi*(r**2)*(R_nl_2(r)**2)))
ax2.set_xlabel("r")
ax2.set_ylabel("$D_{nl}(r)$")

ax3.plot(r,4*np.pi*(r**2)*(R_nl_3(r)**2),label="2p")
ax3.set_title("2p",fontweight='bold')
ax3.set_xlim(0,70)
ax3.set_ylim(0,max(4*np.pi*(r**2)*(R_nl_3(r)**2)))
ax3.set_xlabel("r")
ax3.set_ylabel("$D_{nl}(r)$")

ax4.plot(r,4*np.pi*(r**2)*(R_nl_4(r)**2),label="3p")
ax4.set_title("3p",fontweight='bold')
ax4.set_xlim(0,70)
ax4.set_ylim(0,max(4*np.pi*(r**2)*(R_nl_4(r)**2)))
ax4.set_xlabel("r")
ax4.set_ylabel("$D_{nl}(r)$")

ax5.plot(r,4*np.pi*(r**2)*(R_nl_5(r)**2),label="3d")
ax5.set_title("3d",fontweight='bold')
ax5.set_xlim(0,70)
ax5.set_ylim(0,max(4*np.pi*(r**2)*(R_nl_5(r)**2)))
ax5.set_xlabel("r")
ax5.set_ylabel("$D_{nl}(r)$")

ax6.plot(r,4*np.pi*(r**2)*(R_nl_6(r)**2),label="4d")
ax6.set_title("4d",fontweight='bold')
ax6.set_xlim(0,70)
ax6.set_ylim(0,max(4*np.pi*(r**2)*(R_nl_6(r)**2)))
ax6.set_xlabel("r")
ax6.set_ylabel("$D_{nl}(r)$")

ax7.plot(r,4*np.pi*(r**2)*(R_nl_7(r)**2),label="4f")
ax7.set_title("4f",fontweight='bold')
ax7.set_xlim(0,70)
ax7.set_ylim(0,max(4*np.pi*(r**2)*(R_nl_7(r)**2)))
ax7.set_xlabel("r")
ax7.set_ylabel("$D_{nl}(r)$")

ax8.plot(r,4*np.pi*(r**2)*(R_nl_8(r)**2),label="5f")
ax8.set_title("5f",fontweight='bold')
ax8.set_xlim(0,70)
ax8.set_ylim(0,max(4*np.pi*(r**2)*(R_nl_8(r)**2)))
ax8.set_xlabel("r")
ax8.set_ylabel("$D_{nl}(r)$")

plt.show()
```
